```python

# VERSION 1 : Calcul direct avec saisie


print("VERSION 1 : CALCUL DES MOYENNES DES UE")

#Calcul pour l'UE 1
print("\n--- UE 1 ---")
nb_notes_ue1 = int(input("Nombre de matières pour l'UE 1 : "))
somme_notes_ue1 = 0
somme_coef_ue1 = 0

for i in range(nb_notes_ue1):
    note = float(input(f"Entrez la note {i+1} : "))
    coef = float(input(f"Entrez le coefficient de cette note : "))
    somme_notes_ue1 += note * coef
    somme_coef_ue1 += coef

ue1 = somme_notes_ue1 / somme_coef_ue1 if somme_coef_ue1 > 0 else 0

#Calcul pour l'UE 2
print("\n--- UE 2 ---")
nb_notes_ue2 = int(input("Nombre de matières pour l'UE 2 : "))
somme_notes_ue2 = 0
somme_coef_ue2 = 0

for i in range(nb_notes_ue2):
    note = float(input(f"Entrez la note {i+1} : "))
    coef = float(input(f"Entrez le coefficient de cette note : "))
    somme_notes_ue2 += note * coef
    somme_coef_ue2 += coef

ue2 = somme_notes_ue2 / somme_coef_ue2 if somme_coef_ue2 > 0 else 0

#Calcul pour l'UE 3
print("\n--- UE 3 ---")
nb_notes_ue3 = int(input("Nombre de matières pour l'UE 3 : "))
somme_notes_ue3 = 0
somme_coef_ue3 = 0

for i in range(nb_notes_ue3):
    note = float(input(f"Entrez la note {i+1} : "))
    coef = float(input(f"Entrez le coefficient de cette note : "))
    somme_notes_ue3 += note * coef
    somme_coef_ue3 += coef

ue3 = somme_notes_ue3 / somme_coef_ue3 if somme_coef_ue3 > 0 else 0

# Affichage des moyennes finales
print("\n--- RÉSULTATS VERSION 1 ---")
print(f"Moyenne UE 1 : {round(ue1, 2)}")
print(f"Moyenne UE 2 : {round(ue2, 2)}")
print(f"Moyenne UE 3 : {round(ue3, 2)}")
```

```python
import numpy as np


# VERSION 2 : Calcul avec des listes


print("VERSION 2 : CALCUL AVEC LISTES ET COEFFICIENTS ENCADRÉS")

#Création de la liste exacte des 21 matières
matiere = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]

# 2. Définition des coefficients pour chaque UE (exactement 21 valeurs chacun)
coef1 = [10, 10, 7, 7, 0, 5, 0, 6, 0, 5, 4, 2, 5, 5, 0, 20, 20, 0, 0, 0, 7]
coef2 = [4, 0, 2, 8, 6, 0, 0, 0, 0, 5, 5, 2, 9, 9, 3, 0, 0, 29, 0, 0, 7]
coef3 = [4, 0, 2, 0, 0, 5, 15, 6, 4, 5, 5, 2, 0, 0, 3, 0, 0, 0, 20, 20, 7]

#Saisie dynamique des notes
notes = []
for elt in matiere:
    val = float(input(f"Quelle est votre note pour {elt} ? "))
    notes.append(val)

# 4. Calcul des moyennes pondérées
UE1 = np.average(notes, weights=coef1)
UE2 = np.average(notes, weights=coef2)
UE3 = np.average(notes, weights=coef3)

# 5. Affichage des résultats
print("\n--- RÉSULTATS VERSION 2 ---")
print(f"Moyenne UE1 : {round(UE1, 2)}")
print(f"Moyenne UE2 : {round(UE2, 2)}")
print(f"Moyenne UE3 : {round(UE3, 2)}")
```

```python
# VERSION 3 : Calcul avec des DICTIONNAIRES


print("VERSION 3 : CALCUL AVEC DICTIONNAIRES")

#Liste des matières du Semestre 1 (R101 à R115 et SAE11 à SAE16)
matieres = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]

#Dictionnaires des coefficients pour chaque UE
dico_coef_ue1 = {
    "R101": 10, "R102": 10, "R103": 7, "R104": 7, "R105": 0, "R106": 5, "R107": 0,
    "R108": 6, "R109": 0, "R110": 5, "R111": 4, "R112": 2, "R113": 5, "R114": 5,
    "R115": 0, "SAE11": 20, "SAE12": 20, "SAE13": 0, "SAE14": 0, "SAE15": 0, "SAE16": 7
}

dico_coef_ue2 = {
    "R101": 4, "R102": 0, "R103": 2, "R104": 8, "R105": 6, "R106": 0, "R107": 0,
    "R108": 0, "R109": 0, "R110": 5, "R111": 5, "R112": 2, "R113": 9, "R114": 9,
    "R115": 3, "SAE11": 0, "SAE12": 0, "SAE13": 29, "SAE14": 0, "SAE15": 0, "SAE16": 7
}

dico_coef_ue3 = {
    "R101": 4, "R102": 0, "R103": 2, "R104": 0, "R105": 0, "R106": 5, "R107": 15,
    "R108": 6, "R109": 4, "R110": 5, "R111": 5, "R112": 2, "R113": 0, "R114": 0,
    "R115": 3, "SAE11": 0, "SAE12": 0, "SAE13": 0, "SAE14": 20, "SAE15": 20, "SAE16": 7
}

#Dictionnaire pour stocker les notes saisies par l'utilisateur
dico_notes = {}

print("\nVeuillez saisir les notes pour chaque ressource/SAÉ :")
for m in matieres:
    note = float(input(f"Note pour {m} : "))
    dico_notes[m] = note

#Fonction pour calculer la moyenne d'une UE à partir des dictionnaires
def calculer_moyenne_ue(notes, coefs):
    somme_notes = 0.0
    somme_coefs = 0.0

    for matiere, coef in coefs.items():
        if coef > 0:
            somme_notes += notes[matiere] * coef
            somme_coefs += coef

    return somme_notes / somme_coefs if somme_coefs > 0 else 0

#Calcul des moyennes pour chaque UE
moy_ue1 = calculer_moyenne_ue(dico_notes, dico_coef_ue1)
moy_ue2 = calculer_moyenne_ue(dico_notes, dico_coef_ue2)
moy_ue3 = calculer_moyenne_ue(dico_notes, dico_coef_ue3)

#Affichage des résultats
print("\n--- RÉSULTATS VERSION 3 ---")
print(f"Moyenne UE1 (Réseaux)       : {round(moy_ue1, 2)}")
print(f"Moyenne UE2 (Télécoms)      : {round(moy_ue2, 2)}")
print(f"Moyenne UE3 (Programmation) : {round(moy_ue3, 2)}")
```

```python
import matplotlib.pyplot as plt


# VERSION 4 : Graphique et Décision du Jury


print("VERSION 4 : GRAPHICS ET DÉCISION DU JURY")

#Liste des matières du Semestre 1
matieres = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]

#Dictionnaires des coefficients par UE
dico_coef_ue1 = {
    "R101": 10, "R102": 10, "R103": 7, "R104": 7, "R105": 0, "R106": 5, "R107": 0,
    "R108": 6, "R109": 0, "R110": 5, "R111": 4, "R112": 2, "R113": 5, "R114": 5,
    "R115": 0, "SAE11": 20, "SAE12": 20, "SAE13": 0, "SAE14": 0, "SAE15": 0, "SAE16": 7
}

dico_coef_ue2 = {
    "R101": 4, "R102": 0, "R103": 2, "R104": 8, "R105": 6, "R106": 0, "R107": 0,
    "R108": 0, "R109": 0, "R110": 5, "R111": 5, "R112": 2, "R113": 9, "R114": 9,
    "R115": 3, "SAE11": 0, "SAE12": 0, "SAE13": 29, "SAE14": 0, "SAE15": 0, "SAE16": 7
}

dico_coef_ue3 = {
    "R101": 4, "R102": 0, "R103": 2, "R104": 0, "R105": 0, "R106": 5, "R107": 15,
    "R108": 6, "R109": 4, "R110": 5, "R111": 5, "R112": 2, "R113": 0, "R114": 0,
    "R115": 3, "SAE11": 0, "SAE12": 0, "SAE13": 0, "SAE14": 20, "SAE15": 20, "SAE16": 7
}

#Saisie des notes
dico_notes = {}
print("\nVeuillez saisir les notes pour chaque ressource/SAÉ :")
for m in matieres:
    note = float(input(f"Note pour {m} : "))
    dico_notes[m] = note

#Fonction de calcul
def calculer_moyenne_ue(notes, coefs):
    somme_notes = sum(notes[m] * c for m, c in coefs.items() if c > 0)
    somme_coefs = sum(c for c in coefs.values() if c > 0)
    return somme_notes / somme_coefs if somme_coefs > 0 else 0

moy_ue1 = round(calculer_moyenne_ue(dico_notes, dico_coef_ue1), 2)
moy_ue2 = round(calculer_moyenne_ue(dico_notes, dico_coef_ue2), 2)
moy_ue3 = round(calculer_moyenne_ue(dico_notes, dico_coef_ue3), 2)

moyennes = [moy_ue1, moy_ue2, moy_ue3]

#Détermination de la décision du jury selon le règlement
# Règle : comptage des UE >= 10
ue_validees = sum(1 for m in moyennes if m >= 10)
ue_entre_8_et_10 = sum(1 for m in moyennes if 8 <= m < 10)
ue_sous_8 = sum(1 for m in moyennes if m < 8)

if ue_validees == 3:
    decision = "Année validée ! Passage en BUT 2 autorisé."
elif ue_validees == 2 and ue_entre_8_et_10 == 1:
    decision = "Année non validée, mais passage en BUT 2 autorisé (UE à valider obligatoirement en BUT 2)."
elif ue_validees == 2 and ue_sous_8 == 1:
    decision = "Année non validée et passage en BUT 2 refusé. Redoublement possible sur décision du conseil."
else:
    decision = "Année non validée et passage en BUT 2 exclu."

print("\n--- RÉSULTATS ET DÉCISION DU JURY ---")
print(f"Moyenne UE 1 : {moy_ue1}")
print(f"Moyenne UE 2 : {moy_ue2}")
print(f"Moyenne UE 3 : {moy_ue3}")
print(f"\nDécision du jury : {decision}")

#Affichage du graphique
ues = ['UE 1 (Réseaux)', 'UE 2 (Télécoms)', 'UE 3 (Prog)']
couleurs = ['green' if m >= 10 else 'orange' if m >= 8 else 'red' for m in moyennes]

plt.figure(figsize=(8, 5))
bars = plt.bar(ues, moyennes, color=couleurs, width=0.5)

#Lignes de référence pour le jury (10 = validation, 8 = compensation)
plt.axhline(y=10, color='blue', linestyle='--', label='Seuil de validation (10)')
plt.axhline(y=8, color='red', linestyle=':', label='Seuil éliminatoire (8)')

# Affichage de la valeur sur chaque barre
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval + 0.3, f"{yval:.2f}", ha='center', va='bottom', fontweight='bold')

plt.ylim(0, 20)
plt.ylabel('Moyenne / 20')
plt.title(f"Moyennes des UE - Semestre 1\nDécision : {decision}", fontsize=10, color='darkblue')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```
